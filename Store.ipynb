```python
input_cols=['Store', 'DayOfWeek', 'Promo',
       'StateHoliday', 'StoreType', 'Assortment',
        'Day', 'Month', 'Year']
```

```python
target_cols='Sales'
```

```python
train_inputs=train_df[input_cols].copy()
train_targets=train_df[target_cols].copy()

train_inputs['StateHoliday'] = train_inputs['StateHoliday'].astype(str)
train_inputs['DayOfWeek'] = train_inputs['DayOfWeek'].astype(str)
train_inputs['Promo'] = train_inputs['Promo'].astype(str)
```

```python
val_inputs=val_df[input_cols].copy()
val_inputs['StateHoliday'] = val_inputs['StateHoliday'].astype(str)
val_inputs['DayOfWeek'] = val_inputs['DayOfWeek'].astype(str)
val_inputs['Promo'] = val_inputs['Promo'].astype(str)
```

```python
val_targets=val_df[target_cols].copy()
```

```python
test_inputs=merge_test_df[input_cols].copy()
test_inputs['StateHoliday'] = test_inputs['StateHoliday'].astype(str)
test_inputs['DayOfWeek'] = test_inputs['DayOfWeek'].astype(str)
test_inputs['Promo'] = test_inputs['Promo'].astype(str)
```

```python
numerical_cols=['Store','Day',"Month","Year"]
categorical_cols=['DayOfWeek', 'Promo',
       'StateHoliday', 'StoreType', 'Assortment']
```

```python
from sklearn.impute import SimpleImputer
```

```python
imputer=SimpleImputer(strategy='mean').fit(train_inputs[numerical_cols])
```

```python
train_inputs[numerical_cols]=imputer.transform(train_inputs[numerical_cols])
```

```python
val_inputs[numerical_cols]=imputer.transform(val_inputs[numerical_cols])
test_inputs[numerical_cols]=imputer.transform(test_inputs[numerical_cols])
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
MinMax=MinMaxScaler().fit(train_inputs[numerical_cols])
```

```python
train_inputs[numerical_cols]=MinMax.transform(train_inputs[numerical_cols])
val_inputs[numerical_cols]=MinMax.transform(val_inputs[numerical_cols])
test_inputs[numerical_cols]=MinMax.transform(test_inputs[numerical_cols])
```

```python
train_inputs
```

```python
categorical_cols
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
OneHotEncoder=OneHotEncoder(handle_unknown='ignore').fit(train_inputs[categorical_cols])
```

```python
encoded_cols=list(OneHotEncoder.get_feature_names_out(categorical_cols))
```

```python
encoded_cols
```

```python
train_inputs[encoded_cols] = OneHotEncoder.transform(train_inputs[categorical_cols]).toarray()
val_inputs[encoded_cols] = OneHotEncoder.transform(val_inputs[categorical_cols]).toarray()
test_inputs[encoded_cols] = OneHotEncoder.transform(test_inputs[categorical_cols]).toarray()
```

```python
train_inputs
```

```python
X_train=train_inputs[encoded_cols+numerical_cols].copy()
X_val=val_inputs[encoded_cols+numerical_cols].copy()
X_test=test_inputs[encoded_cols+numerical_cols].copy()
```

```python
from sklearn.linear_model import LinearRegression
```

```python
linear_req=LinearRegression().fit(X_train,train_targets)
```

```python
linear_req.score(X_train,train_targets)
```

```python
trains_pred = linear_req.predict(X_train)
```

```python
val_pred=linear_req.predict(X_val)
```

```python
from sklearn.metrics import mean_squared_error
```

```python
import numpy as np

np.sqrt(mean_squared_error(train_targets, trains_pred))
```

```python
np.sqrt(mean_squared_error(val_targets, val_pred))
```

```python
def try_model(model):
  model.fit(X_train,train_targets)
  train_pred=model.predict(X_train)
  val_pred=model.predict(X_val)
  train_rmse=np.sqrt(mean_squared_error(train_targets,train_pred))
  val_rmse=np.sqrt(mean_squared_error(val_targets,val_pred))
  return train_rmse,val_rmse
```

```python
try_model(LinearRegression())
```

```python
from sklearn.linear_model import Ridge,Lasso,ElasticNet,SGDRegressor
```

```python
try_model(Ridge(alpha=0.8))
```

```python
try_model(Lasso(alpha=0.8))

```

```python
try_model(ElasticNet())
```

```python
try_model(DecisionTreeRegressor())
```
